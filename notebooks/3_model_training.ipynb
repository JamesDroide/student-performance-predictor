```python
import joblib
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from src.data_preprocessing import cargar_y_preprocesar_datos
```

```python
# 1. Cargar y preprocesar los datos
ruta_csv = r'C:\Users\james\OneDrive\Documentos\PREGRADO UPAO\CICLO 9\Machine Learnig\Proyecto\data\raw\StudentPerformanceFactors.csv'
df = cargar_y_preprocesar_datos(ruta_csv, normalizar=True)
```

```python

# Definir la función para clasificar el rendimiento académico
def classify_performance(score):
    if score < 13:
        return 'bajo'
    elif 13 <= score < 14:
        return 'medio'
    else:
        return 'alto'
df.head(10)
```

```python
# Importar librerías necesarias
# Paso 1: Definir las características (X) y la variable objetivo (y)
X = df.drop(columns=['calificacion_examen'])
y = df['calificacion_examen'].apply(classify_performance)  # Función para clasificar el rendimiento
```
