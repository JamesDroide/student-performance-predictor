```python
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_preprocessing import cargar_y_preprocesar_datos
```

```python
# 1. Configuración visual
plt.style.use("ggplot")
sns.set(font_scale=1.1)
```

```python
# 2. Cargar los datos y el modelo entrenado
ruta_csv = '../data/raw/StudentPerformanceFactors.csv'
df = cargar_y_preprocesar_datos(ruta_csv, normalizar=True)
```

```python
# Selección de columnas numéricas
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
```

```python
# 3. Graficar distribuciones
for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribución de {col}')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.tight_layout()
    plt.show()
```

```python
# 4. Boxplots por género usando la columna genero_Male
for col in num_cols:
    plt.figure(figsize=(12, 6))

    # Gráfico para los masculinos (1 en genero_Male)
    plt.subplot(1, 2, 1)
    sns.boxplot(data=df[df['genero_Male'] == 1], x='genero_Male', y=col)
    plt.title(f'{col} para Género Masculino')
    plt.xlabel('Masculino')
    plt.ylabel(col)

    # Gráfico para las féminas (0 en genero_Male)
    plt.subplot(1, 2, 2)
    sns.boxplot(data=df[df['genero_Male'] == 0], x='genero_Male', y=col)
    plt.title(f'{col} para Género Femenino')
    plt.xlabel('Femenino')
    plt.ylabel(col)

    # Ajustar la distribución y los márgenes
    plt.tight_layout()
    plt.show()
```

```python
# Matriz de correlación
plt.figure(figsize=(12, 8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlación")
plt.tight_layout()
plt.show()
```

```python
# Boxplot para detección de outliers
for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=df, y=col)
    plt.title(f'Detección de outliers en {col}')
    plt.tight_layout()
    plt.show()
```

```python
# Scatterplot entre horas de estudio y calificación
plt.figure(figsize=(8, 5))
sns.scatterplot(x='horas_estudio', y='calificacion_examen', data=df)
plt.title("Relación entre Horas de Estudio y Calificación del Examen")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación del Examen")
plt.tight_layout()
plt.show()
```
