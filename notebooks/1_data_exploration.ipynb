```
# 1. Importación de librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```
# Configuración visual
plt.style.use("ggplot")
sns.set(font_scale=1.1)
```

```
# 2. Carga de datos
data_path = 'C:\\Users\\james\\OneDrive\\Documentos\\PREGRADO UPAO\\CICLO 9\\Machine Learnig\\Proyecto\\data\\raw\\StudentPerformanceFactors.csv'
df = pd.read_csv(data_path)
```

```
# 3. Inspección inicial
print("Dimensiones:", df.shape)
print("\nPrimeras filas:")
df.head()
```

```
print("\nTipos de datos:")
df.info()

print("\nValores nulos:")
df.isnull().sum()
```

```
# Visualización de valores nulos
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='coolwarm')
plt.title("Visualización de Valores Nulos")
plt.tight_layout()
plt.show()
```

```
# Transformar notas de 0–100 a 0–20
if 'Exam_Score' in df.columns:
    df['Exam_Score'] = df['Exam_Score'] * 0.2

if 'Previous_Scores' in df.columns:
    df['Previous_Scores'] = df['Previous_Scores'] * 0.2
```

```
# 5. Análisis de distribución de variables numéricas
num_cols = df.select_dtypes(include=['float64', 'int64']).columns

for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribución de {col}')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
    plt.tight_layout()
    plt.show()
```

### Distribución de Variables

1. **Distribución de Horas de Estudio (Hours_Studied)**

2. **Distribución de Asistencia (Attendance)**
   La distribución de asistencia es bastante equilibrada, con picos en el rango del 70% al 80% de asistencia. A medida que nos acercamos al 100%, la cantidad de estudiantes con asistencia disminuye. 3. **Distribución de Horas de Sueño (Sleep_Hours)**
   La distribución de las horas de sueño revela picos en valores de 6, 7 y 8 horas de sueño, lo que sugiere que la mayoría de los estudiantes duermen entre 6 y 8 horas. Este patrón es común entre los estudiantes que suelen equilibrar sus estudios con el descanso.

4. **Distribución de Calificaciones Previas (Previous_Scores)**
   La distribución de las calificaciones previas es más uniforme, con una ligera tendencia a concentrarse alrededor de las calificaciones entre 12 y 16. Esta distribución indica que muchos estudiantes se encuentran en un rango medio de desempeño académico.

5. **Distribución de Sesiones de Tutoría (Tutoring_Sessions)**
   La distribución de las sesiones de tutoría muestra que la mayoría de los estudiantes no asisten a muchas sesiones de tutoría. Sin embargo, algunos estudiantes han asistido a entre 1 y 2 sesiones, y un número reducido de estudiantes tiene más sesiones.

6. **Distribución de Actividad Física (Physical_Activity)**
   Los estudiantes muestran una concentración de actividad física en valores de 2 a 4, lo que sugiere que la mayoría realiza actividad física de moderada a intensa. Esto es consistente con la importancia de la actividad física para el bienestar general.

7. **Distribución de Calificación del Examen (Exam_Score)**
   La distribución de las calificaciones de examen muestra una forma ligeramente sesgada hacia la derecha, con un pico alrededor de 13. Esto indica que la mayoría de los estudiantes obtuvo calificaciones cercanas a la media, con algunos estudiantes que tuvieron un desempeño superior.

```
# 7. Correlación entre variables
plt.figure(figsize=(12, 8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlación")
plt.tight_layout()
plt.show()
```

# Matriz de Correlación

La **matriz de correlación** muestra las relaciones entre las variables en el conjunto de datos. Se destacan las siguientes correlaciones:

1. **Horas de Estudio (Hours_Studied) y Calificación del Examen (Exam_Score)**
   Existe una correlación positiva moderada (0.45) entre las horas de estudio y la calificación del examen. Esto sugiere que a medida que los estudiantes dedican más tiempo al estudio, es probable que sus calificaciones en el examen mejoren.

2. **Asistencia (Attendance) y Calificación del Examen (Exam_Score)**
   La correlación entre la asistencia y la calificación del examen es más fuerte (0.58), lo que indica que los estudiantes con una mayor asistencia a clase tienden a obtener mejores resultados en los exámenes.

3. **Horas de Estudio (Hours_Studied) y Asistencia (Attendance)**
   La correlación entre horas de estudio y asistencia es prácticamente nula (-0.01), lo que sugiere que no hay una relación clara entre la cantidad de tiempo dedicado al estudio y la asistencia a clases.

4. **Horas de Sueño (Sleep_Hours) y Calificación del Examen (Exam_Score)**
   La correlación entre las horas de sueño y las calificaciones del examen es muy baja (-0.02), indicando que no existe una relación significativa entre el número de horas de sueño y el rendimiento en el examen.

5. **Horas de Estudio (Hours_Studied) y Horas de Sueño (Sleep_Hours)**
   Hay una correlación ligera y positiva (0.01) entre las horas de estudio y las horas de sueño, lo que sugiere que, aunque es una relación débil, los estudiantes que estudian más también tienden a dormir ligeramente más.

6. **Sesiones de Tutoría (Tutoring_Sessions) y Calificación del Examen (Exam_Score)**
   La correlación entre las sesiones de tutoría y las calificaciones del examen es baja (0.16), indicando que aunque las tutorías pueden tener algo de impacto, no son un factor decisivo para las calificaciones en el examen.

7. **Actividad Física (Physical_Activity) y Calificación del Examen (Exam_Score)**
   La correlación entre la actividad física y la calificación del examen es muy baja (0.03), lo que sugiere que la actividad física no tiene un impacto considerable en los resultados del examen.



```
# 8. Relación entre horas de estudio y calificación del examen
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Hours_Studied', y='Exam_Score', data=df)
plt.title("Relación entre Horas de Estudio y Calificación del Examen")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación del Examen (0-20)")
plt.tight_layout()
plt.show()
```

# Relación entre Horas de Estudio y Calificación del Examen

Este **gráfico de dispersión** muestra la relación entre las **Horas de Estudio** y las **Calificaciones del Examen**. Observamos que:

1. **Relación positiva**: Existe una clara **tendencia positiva** en la que, a medida que aumentan las horas de estudio, las calificaciones del examen tienden a aumentar. Este patrón sugiere que los estudiantes que dedican más tiempo al estudio suelen obtener mejores resultados en sus exámenes.

2. **Distribución dispersa**: Aunque la mayoría de los puntos se agrupan en la parte inferior, indicando calificaciones más bajas para menos horas de estudio, hay una **dispersión creciente** a medida que las horas de estudio aumentan, lo que refleja una mayor variabilidad en las calificaciones a niveles de estudio más altos.

3. **Outliers**: Se observan algunos **puntos fuera de la tendencia principal** que pueden representar estudiantes que, aunque estudiaron mucho, no lograron mejorar significativamente su calificación, o aquellos que, a pesar de estudiar poco, obtuvieron buenos resultados.
