# Predicción de rendimiento académico de estudiantes

**Problematica:**
En las instituciones educativas, uno de los mayores desafíos es identificar a los estudiantes que están en riesgo de bajo rendimiento académico antes de que sea demasiado tarde. Sin herramientas adecuadas para monitorear de forma preventiva el desempeño de los estudiantes, los esfuerzos de intervención suelen ser reactivos, lo que aumenta la tasa de deserción estudiantil.

**Obejtivo del Proyecto:**
Este proyecto tiene como objetivo predecir el rendimiento académico de los estudiantes mediante técnicas de Machine Learning, específicamente con el modelo Random Forest. Utilizando datos históricos sobre factores como horas de estudio, asistencia, nivel de motivación, acceso a recursos educativos, calidad docente, actividad física, género y calificaciones previas, el sistema predice el rendimiento académico futuro de los estudiantes.

**Hipótesis:**
"Se espera que el rendimiento académico de los estudiantes esté determinado por una combinación de factores como la asistencia a clases, las horas de estudio, el acceso a recursos educativos, la calidad docente, el nivel de motivación, la actividad física, las calificaciones previas y el género."

**Datos de entrada:**

---
**Horas de estudio:** Número de horas de esudio que el estudiante que dedica por semana.
**Asistencia:** Frecuencia con la que el estudiante asiste a las clases.
**Acceso a recursos educativos:** Disponibilidad de materiales, tecnología y apoyo educativo.
**Nivel de motivación:** Medición subjetiva de la motivación del estudiante hacia sus estudios.
**Calidad docente:** Valoración del desempeño y habilidad de los docentes para enseñar.
**Actividad física:** Participación en actividades físicas o deportivas, considerada un factor que influye en el bienestar y el rendimiento cognitivo.
**Género:** Característica demográfica que puede ser un factor a considerar en análisis de datos académicos.
**Calificación del examen:** Resultado del examen final o más relevante, utilizado como indicador del rendimiento previo del estudiante.


**Datos de Salida:**
La variable de salida corresponde a la predicción del rendimiento académico del estudiante, representado por una calificación numérica en una escala de **0 a 20**.

**Variable:** Cuantitativo

**Rubro Dirigido:**
Educación: Instituciones educativas (escuelas, colegios).

**Usuarios Finales:**

---
**Profesores:** Los profesores utilizarán el sistema para obtener informes detallados sobre el rendimiento de los estudiantes. A partir de estos informes, podrán identificar estudiantes con bajo rendimiento y tomar decisiones de intervención.
**Directores:** Los directores accederán a reportes ejecutivos que les permitirán gestionar y tomar decisiones a nivel institucional basadas en las tendencias de rendimiento académico.
